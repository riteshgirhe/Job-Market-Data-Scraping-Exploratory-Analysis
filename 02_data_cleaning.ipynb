```python
df["title"]=df["title"].str.strip().str.title()
df["company"]=df["company"].str.strip().str.title()
df["location"]=df["location"].str.strip().str.title()
df["location"]=df["location"].str.strip()
df["posted_text"]=df["posted_text"].str.strip().str.title()
df["posted_date"]=pd.to_datetime(df["posted_date"] , errors="coerce")
```

```python
df["experience"] = df["experience"].fillna("Unknown")
df["experience"] = df["experience"].str.strip().str.title()

#order = ["Entry Level", "Associate", "Mid-Senior Level", "Not Applicable", "Unknown"]
#df["experience"] = pd.Categorical(df["experience"], categories=order, ordered=True)

df["experience"]=df["experience"].str.replace("Not Applicable","Experience").str.replace("Unknown","Entry Level")
```

```python
df["experience"].unique()
df["experience"].value_counts()
df["experience"].isna().sum()
df["experience"].value_counts(dropna=False)
df["experience"].value_counts()
```

```python
df["salary_min_new"] = df["salary_min"] * 100000
df["salary_min_new"] = df["salary_min_new"] / (100 * 12)
df[["salary_min_new", "salary_max"]].head(20)
df["Salary"] = (df["salary_min_new"] + df["salary_max"]) / 2
df["Salary1"] = (df["salary_min"] + df["salary_max"]) / 2
```

```python
df.drop(["salary_min", "salary_max", "salary_min_new","salary", "Salary1"], axis=1, inplace=True)
```

```python
df["job_type"]=df["job_type"].fillna("Unknown")
df["job_type"]=df["job_type"].str.strip().str.title().replace("Fulltime","Full Time").replace("Unknown","Intern")
```

```python
df["skills"] = df["skills"].str.strip().str.title()

df = df.dropna(subset=["skills"])

print(df["skills"].isna().sum())

df["skills_list"] = df["skills"].str.split(",")

df["skills_list"] = df["skills_list"].apply(
    lambda x: [skill.strip() for skill in x] if isinstance(x, list) else x
)

from collections import Counter

skill_counts = Counter(
    skill
    for skills in df["skills_list"].dropna()
    for skill in skills
)

skill_counts.most_common(10)
```

```python
df.columns = df.columns.str.strip().str.title()
```
