```python
def scrape_page(start):

    url = f"https://www.linkedin.com/jobs/search/?keywords=ai%20and%20data&location=India&start={start}"

    response = requests.get(url, headers=headers)

    print("Start:", start, "Status:", response.status_code)

    soup = BeautifulSoup(response.text, "html.parser")

    job_cards = soup.find_all(
        "div",
        class_="base-search-card"
    )

    page_jobs = []

    for card in job_cards:

        title_tag = card.find(
            "h3",
            class_="base-search-card__title"
        )

        company_tag = card.find(
            "h4",
            class_="base-search-card__subtitle"
        )

        location_tag = card.find(
            "span",
            class_="job-search-card__location"
        )

        link_tag = card.find(
            "a",
            href=lambda x: x and "/jobs/view/" in x
        )

        date_tag = card.find("time")

        page_jobs.append({
            "title": title_tag.get_text(strip=True) if title_tag else None,
            "company": company_tag.get_text(strip=True) if company_tag else None,
            "location": location_tag.get_text(strip=True) if location_tag else None,
            "job_url": link_tag.get("href") if link_tag else None,
            "posted_text": date_tag.get_text(strip=True) if date_tag else None,
            "posted_date": date_tag.get("datetime") if date_tag else None
        })

    return page_jobs
```

```python
def scrape_page(start):

    url = (
        "https://www.linkedin.com/jobs/search/"
        "?keywords=ai%20and%20data"
        "&location=India"
        f"&start={start}"
    )

    response = requests.get(
        url,
        headers=headers,
        timeout=20
    )

    print("Start:", start, "Status:", response.status_code)

    if response.status_code == 429:
        print("Rate limit reached. Stopping requests.")
        return None

    if response.status_code != 200:
        print("Request failed.")
        return None

    soup = BeautifulSoup(response.text, "html.parser")

    job_cards = soup.find_all(
        "div",
        class_="base-search-card"
    )

    page_jobs = []

    for card in job_cards:

        title_tag = card.find(
            "h3",
            class_="base-search-card__title"
        )

        company_tag = card.find(
            "h4",
            class_="base-search-card__subtitle"
        )

        location_tag = card.find(
            "span",
            class_="job-search-card__location"
        )

        link_tag = card.find(
            "a",
            href=lambda x: x and "/jobs/view/" in x
        )

        date_tag = card.find("time")

        page_jobs.append({
            "title": title_tag.get_text(strip=True) if title_tag else None,
            "company": company_tag.get_text(strip=True) if company_tag else None,
            "location": location_tag.get_text(strip=True) if location_tag else None,
            "job_url": link_tag.get("href") if link_tag else None,
            "posted_text": date_tag.get_text(strip=True) if date_tag else None,
            "posted_date": date_tag.get("datetime") if date_tag else None
        })

    return page_jobs
```

```python
def extract_criteria(soup):

    experience = None
    job_type = None

    criteria_items = soup.find_all(
        "li",
        class_="description__job-criteria-item"
    )

    for item in criteria_items:

        text = item.get_text(" ", strip=True)

        if "Seniority level" in text:
            value = item.find(
                "span",
                class_="description__job-criteria-text"
            )

            if value:
                experience = value.get_text(strip=True)

        elif "Employment type" in text:
            value = item.find(
                "span",
                class_="description__job-criteria-text"
            )

            if value:
                job_type = value.get_text(strip=True)

    return experience, job_type
```

```python
import json
import re

def extract_job_details(soup):

    experience = None
    job_type = None
    salary = None
    skills = []

    # -------------------------
    # 1. Experience + Job Type
    # -------------------------

    criteria_items = soup.find_all(
        "li",
        class_="description__job-criteria-item"
    )

    for item in criteria_items:

        text = item.get_text(" ", strip=True)

        value = item.find(
            "span",
            class_="description__job-criteria-text"
        )

        if not value:
            continue

        value = value.get_text(strip=True)

        if "Seniority level" in text:
            experience = value

        elif "Employment type" in text:
            job_type = value

    # -------------------------
    # 2. JSON-LD data
    # -------------------------

    json_ld = soup.find(
        "script",
        type="application/ld+json"
    )

    data = {}

    if json_ld:
        try:
            data = json.loads(json_ld.string)
        except:
            data = {}

    description = data.get("description", "")

    # -------------------------
    # 3. Salary
    # -------------------------

    salary_tag = soup.find(
        class_=re.compile("salary", re.I)
    )

    if salary_tag:
        salary = salary_tag.get_text(" ", strip=True)

    # -------------------------
    # 4. Skills
    # -------------------------

    skill_list = [
        "Python",
        "SQL",
        "R",
        "Java",
        "C++",
        "JavaScript",
        "Machine Learning",
        "Deep Learning",
        "Artificial Intelligence",
        "NLP",
        "TensorFlow",
        "PyTorch",
        "Keras",
        "Hugging Face",
        "Pandas",
        "NumPy",
        "Scikit-learn",
        "Matplotlib",
        "Power BI",
        "Tableau",
        "AWS",
        "Azure",
        "GCP",
        "Docker",
        "Kubernetes",
        "Spark",
        "Databricks",
        "MongoDB",
        "PostgreSQL",
        "Git"
    ]

    for skill in skill_list:

        if re.search(
            r"\b" + re.escape(skill) + r"\b",
            description,
            re.IGNORECASE
        ):
            skills.append(skill)

    return {
        "experience": experience,
        "salary": salary,
        "skills": ", ".join(skills) if skills else None,
        "job_type": job_type
    }
```

```python
def get_job_details(job_url):

    try:
        response = requests.get(
            job_url,
            headers=headers,
            timeout=20
        )

        if response.status_code != 200:
            print(
                "Failed:",
                response.status_code,
                job_url
            )
            return {
                "experience": None,
                "salary": None,
                "skills": None,
                "job_type": None
            }

        soup = BeautifulSoup(
            response.text,
            "html.parser"
        )

        return extract_job_details(soup)

    except Exception as e:

        print("Error:", e)

        return {
            "experience": None,
            "salary": None,
            "skills": None,
            "job_type": None
        }
```

```python
def get_job_details(job_url):

    try:
        response = requests.get(
            job_url,
            headers=headers,
            timeout=20
        )

        if response.status_code == 429:
            print("429 RATE LIMIT — STOPPING")
            return None

        if response.status_code != 200:
            return {
                "experience": None,
                "salary": None,
                "skills": None,
                "job_type": None
            }

        soup = BeautifulSoup(
            response.text,
            "html.parser"
        )

        return extract_job_details(soup)

    except Exception as e:
        print("Error:", e)

        return {
            "experience": None,
            "salary": None,
            "skills": None,
            "job_type": None
        }
```

```python
import re

def extract_salary(salary):

    if pd.isna(salary):
        return pd.Series([None, None])

    numbers = re.findall(
        r"[\d,]+(?:\.\d+)?",
        str(salary)
    )

    numbers = [
        float(x.replace(",", ""))
        for x in numbers
    ]

    if len(numbers) >= 2:
        return pd.Series([numbers[0], numbers[1]])

    elif len(numbers) == 1:
        return pd.Series([numbers[0], numbers[0]])

    return pd.Series([None, None])
```

```python
df[["salary_min", "salary_max"]] = df["salary"].apply(
    extract_salary
)
```

```python
from collections import Counter

all_skills = []

for skills in df["skills"].dropna():
    for skill in skills.split(","):
        all_skills.append(skill.strip())

skill_counts = Counter(all_skills)

print(skill_counts.most_common(20))
```
